# 📦 E-Commerce-Retouren-Ampel

Der Online-Handel in Deutschland ist Weltmeister im Bestellen – aber leider auch im Zurückschicken. Während globale Kennzahlen oft die ökonomischen Verluste beleuchten, bleibt der ökologische Fußabdruck und der operative Aufwand für den einzelnen Händler oft abstrakt. Die „E-Commerce-Retouren-Ampel“ setzt genau hier an. Das Projekt übersetzt komplexe Nutzer- und Produktdaten in ein einfaches, visuelles Signal: Eine Ampel, die das Retourenrisiko eines Warenkorbs in Echtzeit bewertet.

## 🔍 Zentrale Forschungsfrage

Die leitende Forschungsfrage lautet:
Wie lassen sich historische Transaktionsdaten und Kundenverhaltensmuster mithilfe von Machine-Learning-Modellen so analysieren und visualisieren, dass sie für Online-Händler eine belastbare Vorhersage über die Retourenwahrscheinlichkeit liefern – und dabei für die operative Entscheidungsfindung (z.B. Logistikplanung) intuitiv nutzbar bleiben?

Im Fokus steht die Übersetzungsleistung: Weg von unübersichtlichen Tabellen hin zu einer klaren, handlungsrelevanten Aussage (Grün/Gelb/Rot)

### Thematische Forschungsfragen

Zur strukturierten Bearbeitung konzentriert sich das Projekt auf drei Perspektiven:

1. Feature-Engineering & Korrelation: Welche Merkmale (z.B. demografische Daten, Tageszeit, Rabatt-Level, Produktkategorie) sind die stärksten Indikatoren für eine Retoure in deutschen Online-Shops?  

2. Modell-Präzision: Wie hoch ist die Treffsicherheit eines binären Klassifikationsmodells bei der Identifikation von „High-Risk“-Bestellungen?  

3. Präventive Wirkung: Inwieweit lässt sich durch die Integration der Vorhersage in das Backend ein „Nudging“-Effekt erzielen, um die Logistikkette zu entlasten?

## Zielgruppe

Die Ergebnisse dieser Analyse richten sich primär an:

* E-Commerce Manager: Zur Sortimentssteuerung und Gewinnmaximierung.

* Qualitätsmanagement: Zur Identifikation von Produktmängeln oder Transportschäden.

* Logistik-Planer: Zur Optimierung von Verpackungsprozessen bei hohen Schadensraten.

## 📏 Zielgrößen und Evaluationsmatriken

Zur Bewertung der Modellleistung werden folgende Metriken herangezogen:

* Precision (Hauptmetrik):
Wie viele als „risikoreich“ klassifizierte Bestellungen sind tatsächlich Retouren?
→ Wichtig, um falsche Alarme zu minimieren

* Recall:
Wie viele tatsächliche Retouren werden erkannt?
→ Wichtig, um möglichst viele problematische Bestellungen zu identifizieren

* F1-Score:
Harmonisches Mittel aus Precision und Recall zur Gesamtbewertung

* Accuracy (untergeordnet):
Gesamtanteil korrekt klassifizierter Fälle

Der Fokus liegt auf einer hohen Precision, da Fehlklassifikationen (z.B. unnötige Eingriffe in den Kaufprozess) vermieden werden sollen.

## ⚙️ Technische Umsetzung & ML-Logik

Das Herzstück der Anwendung ist ein strukturierter Machine-Learning-Workflow, der mehrere aufeinander aufbauende Schritte umfasst.

Die Modelllogik folgt einem Supervised Learning Ansatz:

* Algorithmen:
Einsatz von Klassifikationsverfahren wie Random Forest und XGBoost, da diese besonders gut mit tabellarischen Daten umgehen können.

* Datenaufbereitung:
Bereinigung der Daten (Missing Values, Encoding, Feature Engineering)


* Validierung:
Training und Test auf historischen Daten zur Sicherstellung der Generalisierbarkeit

* Output:
Wahrscheinlichkeiten (0–1), die in Ampelwerte übersetzt werden
(z.B. >70% = 🔴)

## 📊 Datengrundlage

Für die Entwicklung der ML-App werden mehrere komplementäre Datensätze verwendet:

* Der **[UCI Online Retail Dataset](https://archive-beta.ics.uci.edu/dataset/352/online+retail)** dient als primäre Trainingsquelle. Retouren sind hier indirekt über Stornierungen erkennbar (InvoiceNo beginnt mit „C“), wodurch eine binäre Zielvariable erstellt werden kann - dies bietet, durch echte Transaktionen auf Einzelartikel-Ebene, das sauberste Fundament für das Erlernen grundlegender Retourenmuster.
  
* Der **[Olist Brazilian E-Commerce Dataset](https://www.kaggle.com/datasets/olistbr/brazilian-ecommerce)** ergänzt die Analyse durch umfangreiche Kunden- und Logistikdaten und eignet sich besonders für das Feature Engineering, da die Verknüpfung von Lieferdistanzen und Produktmaßen die Identifikation komplexer Risikofaktoren ermöglicht.

* Optional wird der **[Google TheLook Ecommerce Dataset](https://www.kaggle.com/datasets/daichiuchigashima/thelook-ecommerce)** eingebunden, um moderne, skalierbare Big-Data-Strukturen abzubilden und die Performance des Modells unter realitätsnahen Cloud-Bedingungen sicherzustellen.

Die Kombination mehrerer Datensätze ermöglicht eine robustere Modellierung und verbessert die Generalisierbarkeit der Ergebnisse.

## ⚠️ Grenzen und Schwächen

* Individuelle Psychologie: Spontane Retourengründe (z.B. „gefällt mir doch nicht“) sind ohne Daten schwer vorhersehbar.  

* Datenaktualität: ML-Modelle basieren auf der Vergangenheit; plötzliche Änderungen im Konsumverhalten (z.B. Trends) werden zeitverzögert gelernt.
